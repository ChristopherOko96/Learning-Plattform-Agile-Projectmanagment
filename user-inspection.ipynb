# Import Required Modules
Import the necessary modules for user information lookup.

```python
import os
import getpass

try:
    import pwd
except ImportError:
    pwd = None

print('Modules imported successfully.')
```

# Get Current User with os.getlogin()
Use os.getlogin() to fetch the current login name from the OS.

```python
try:
    os_login = os.getlogin()
except Exception as err:
    os_login = f'Error: {err}'

print('os.getlogin():', os_login)
```

# Get Current User with getpass.getuser()
Use getpass.getuser() as an alternative method to determine the current user.

```python
try:
    getpass_user = getpass.getuser()
except Exception as err:
    getpass_user = f'Error: {err}'

print('getpass.getuser():', getpass_user)
```

# Use pwd on Unix to Resolve User Info
Use the pwd module on Unix-like systems to retrieve detailed information about the current user.

```python
if pwd is not None:
    try:
        user_info = pwd.getpwnam(getpass_user)
        print('pwd user info:')
        print('  name:', user_info.pw_name)
        print('  uid:', user_info.pw_uid)
        print('  gid:', user_info.pw_gid)
        print('  home:', user_info.pw_dir)
        print('  shell:', user_info.pw_shell)
    except Exception as err:
        print('pwd lookup error:', err)
else:
    print('pwd module is not available on this platform.')
```

# Display Current User in Jupyter Output
Print the retrieved user information and handle cases where methods may fail.

```python
print('\nSummary of current user information:')
print('os.getlogin() ->', os_login)
print('getpass.getuser() ->', getpass_user)
if pwd is not None and 'user_info' in globals():
    print('pwd information available.')
else:
    print('pwd information not available or not resolved.')
```
